```
import numpy as np
from sympy import symbols, Eq, solve

#Declaration variable
def resolve_equation():

  #Definition symboles
  a, b, c = symbols('Pe, Pm, Pf')

  #Defnition equations
  eq1 = Eq(2*a + 3*b + 2*c, 0)
  eq2 = Eq(a - b + c, 0)
  eq3 = Eq(2*a + 2*b - 3*c, 0)

  #Resollutions
  solutions = solve((eq1, eq2, eq3), (a,b,c), dict=True)

  #Affichage resultats
  if solutions:
    for sol in solutions:
      print(f"Pe = {sol[a]}")
      print(f"Pm = {sol[b]}")
      print(f"Pf= {sol[c]}")
    else:
      print("Pas de solution ou solution infinie")

if __name__ == "__main__":
    resolve_equation()
```
