##Etape 1 : Coordonnées GPS à déchiffrer

Le **chiffrement de César** est un chiffrement dans lequel chaque lettre du texte en clair est remplacée par une autre lettre située à une position fixe dans l'alphabet : cela est appelé le **décalage**.

#Démonstration scientifique

Le chiffrement de Césear repose sur 2 formules :     

- La formule pour le chiffrement : E(x) = (x+n) mod10  
- La formule pour le déchiffrement : D(x) = (x − n) mod10  

Oú :     

- x représente la position de la lettre dans l'alphabet (A=0, B=1, ..., Z=25) ou le chiffre (0, 1, 2, ..., 9)

- n est la clé de décalage

- mod10 est l'opération modulo 10 (vu qu'il y a 10 chiffres : 0 --> 9), qui fait "boucler" les valeurs dans l'intervalle

*Message chiffré* : 9153787770964

*Décalage* : 4

**Chiffrement de Ceasar avec un décalage de 4**

*Coordonnées GPS Déchiffrées* : **57°19'34.3"N 3°36'52.0"W**

Localisation:
[Google Maps](https://www.google.com/maps/place/57%C2%B019'34.3%22N+3%C2%B036'52.0%22W/@57.3007636,-4.5424845,8.29z/data=!4m4!3m3!8m2!3d57.3261944!4d-3.6144444?entry=ttu&g_ep=EgoyMDI1MTAxNC4wIKXMDSoASAFQAw%3D%3D)

```python
def decrypt_coord_GPS(numbers, shift):

  decrypted_numbers = ""
  if len(numbers)!=13:
    return f"Pas le format de coordonnées GPS, 13 chiffres attendus et non {len(numbers)} chiffres :("

  for number in numbers:
    original_number = int(number)
    decrypt_number = (original_number - shift) % 10
    decrypted_numbers += str(decrypt_number)

  print("Nombres chiffrées : " + numbers)
  print()
  return decrypted_numbers


resultats = decrypt_coord_GPS("9153787770964", 4) #Entrez ici la suite de chiffre à dechiffrer et son décalage
print("Nombres déchiffrés : "+ resultats)
print()

def format_gps_coordinates(numbers):

  # Latitude (7 chiffres): XX°XX'XX.X"N
    lat_degrees = numbers[0:2]      # 2 chiffres pour les degrés
    lat_minutes = numbers[2:4]      # 2 chiffres pour les minutes
    lat_seconds = numbers[4:6]      # 2 chiffres pour les secondes
    lat_decimal = numbers[6]        # 1 chiffre pour la décimale

    # Agrégation pour la latitude
    latitude = f"{lat_degrees}°{lat_minutes}'{lat_seconds}.{lat_decimal}\"N"

    # Longitude (6 chiffres): X°XX'XX.X"W
    lon_degrees = numbers[7]        # 1 chiffre pour les degrés (0-9)
    lon_minutes = numbers[8:10]     # 2 chiffres pour les minutes
    lon_seconds = numbers[10:12]    # 2 chiffres pour les secondes
    lon_decimal = numbers[12]       # 1 chiffre pour la décimale

    # Agrégation pour la logitude
    longitude = f"{lon_degrees}°{lon_minutes}'{lon_seconds}.{lon_decimal}\"W"

    # Coordonnées finaux
    return f"{latitude} {longitude}"

formated_gps_coodinates = format_gps_coordinates(resultats)
print("Coordonnées GPS déchiffrées : " + formated_gps_coodinates)
```

##Etape 2 : Message à déchiffrer (ASCII) en utilisant RSA

Le **RSA** (Rivest-Shamir-Adleman) est un algorithme de cryptographie à clé publique inventé en 1977. Contrairement au chiffrement de César qui utilise la même clé pour chiffrer et déchiffrer (cryptographie symétrique), RSA utilise deux clés différentes : une **clé publique** pour chiffrer et une **clé privée** pour déchiffrer.

N = 3233

E = 17

###Les nombres premiers

Un nombre premier est un entier strictement supérieur à 1 qui n’a que deux diviseurs entiers positifs : 1 et lui-même.

###Démonstration scientifique du RSA


1) On initialise nos attributs:
- C : Message encodé

```python
N=3233 # Modulus N
e=17 #Exposant public e

#Message en ASCII à déchiffrer
C = [2726, 1313, 1992, 884, 2412, 1453, 1230, 2185, 2412, 1992,
     1313, 1230, 884, 1992, 281, 1632, 281, 2170, 1453, 1992,
     1230, 2185, 2160, 1230, 1992, 745, 1632, 1992, 612, 745, 1632,
     1627, 2160, 1313, 1992, 2412, 2185, 2160, 2923, 1313]
```

###2. Factorisation de N

Tout d'abord, il faut trouver *p* et *q*, les deux facteurs premiers de *N* grâce à la factorisation :    

*N* = 3233 = *p* x *q*

En testant les diviseus grâce à une boucle *for* dans un script Python, nous trouvons :    

*p* = 53 et *q* = 61

```python
def factoriser_N(N):
  for i in range(2, int(N**0.5) + 1):
    if N % i == 0:
      return i, N // i
  return None, None

two_factors = factoriser_N(N)
print(f"Facteurs trouvés : p = {two_factors[0]}, q = {two_factors[1]}")
```

##3. Calcul de φ(N)

Vu que maintenant on connaît *p* et *q*,on va pouvoir calculer la **fonction totient d'Euclide**, permettant de savoir combien y a-t-il de nombres premiers avec N:

*ϕ*(N) = (p−1) × (q−1) oú *N* = 3233, *p* = 53 et *q* = 61

*ϕ*(3233) = (53−1) × (61−1) = 52 × 60 = 3120

Il y a 3120 nombres premiers avec *N*.

```python
def calcul_phi(p, q):
  return (p - 1)*(q - 1)

phi_N = calcul_phi(two_factors[0], two_factors[1])
print(f"φ(N) = {phi_N}")
```

##3. Algorithme étendu d'Euclide + Calcul Clé Privé
Maintenant, il faudra utiliser l'**algorithme étendu d'Euclide** pour trouver la clé privé *D*.

On va calculer l'**inverse modulaire** avec la relation:   

D × E = 1 (mod ϕ(N))

D × 17 = 1 (mod 3120)

Pour cela, on va utiliser l'équation suivante :     

17 × D + 3120 × k = 1

En appliquant la méthode de substitution inverse, nous obtenons:

D = 2753

```python
def calcul_euclide(a,b):
  if a == 0:
    return b, 0, 1
  pgcd, x1, y1 = calcul_euclide(b % a, a)
  x = y1 - (b // a)* x1
  y = x1
  return pgcd, x, y

euclide = calcul_euclide(e, phi_N)
print(f"PGCD = {euclide[0]}, x = {euclide[1]}, y = {euclide[2]}")
```

```python
def calcul_private_key(e, phi_N):
  pgcd, x, y = euclide
  if  pgcd != 1:
    raise ValueError("e et φ(N) ne sont pas premiers entre eux!")
  return (x % phi_N + phi_N) % phi_N

private_key = calcul_private_key(e, phi_N)
print(f"D = {private_key}")
```

##4. Déchiffrement des valeurs

Maintenant que nous avons la clé privé, on peut passer au décryptage du message. Pur cela, on va utiliser la formule suivante :    

M= C^D mod N

Pour le chiffrement, c'est cette formule :    

C = M^e mod N

Grâce à cela, on pourra déchiffrer notre message oú nous obtenons :    

**Le trésor est caché sous la plaque rouge**

```python
def decrypt_rsa(c, d, n):
  m = pow(c, d, n)
  return m
example_decryptage = decrypt_rsa(C[0], private_key, N)
print(f"Exemple de déchiffrement (Format ASCII) : {example_decryptage}")
```

```python
def decrypt_message(msg, d, n):
    message = ''
    for c in msg:
        m = decrypt_rsa(c, d, n)
        message += chr(m)
    return message
tresor = decrypt_message(C, private_key, N)
print(f"Message décrypté : {tresor}")
```
